```python
from autograd.nodes import Variable
from autograd.utils.visualize import Visualizer
import plotly.io as pio
pio.renderers.default = 'notebook' # o también puedes probar con 'iframe'
```

```python
x = Variable(1.2, label = "x")
y = Variable(2, label = "y")
z = Variable(-2, label = "z")
w = Variable(3.123, label = "w")
s = Variable(6.2351, label =  "s")

a = 2*x**y 
a._label = "a"
b = z * (s / w)  * a
b._label = "b"
c = b - s**(1/2)
c._label = "c"
```

```python
c.zero_grad()
c.backward()
print("c value:\n",c.value, "\n", 
      "grads:\n", 
      "c:", c.grad, ", b:", b.grad, ", a:", a.grad,
      ",\n w:", w.grad, ", z:", z.grad, ", x:", x.grad, 
      ", y:", y.grad, 
      sep="") 
```

```python
Visualizer(c).draw()
```
